
Tests the ~27-day solar rotational periodicity in GCR flux and tropical convective OLR against Monte Carlo AR(1) red-noise surrogates.

```python
import pandas as pd
import matplotlib.pyplot as plt
coh = pd.read_csv('../data/processed/red_noise_coherence_results.csv')
coh[(coh['period_days'] >= 22) & (coh['period_days'] <= 32)]
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
sub = coh[coh['period_days'] <= 65]
ax.plot(sub['period_days'], sub['coherence'], color='teal', lw=1.8, label='Observed GCR-OLR Coherence')
ax.plot(sub['period_days'], sub['red_noise_95'], '--', color='orange', label='95% AR(1) Red-Noise Null')
ax.plot(sub['period_days'], sub['red_noise_99'], ':', color='red', label='99% AR(1) Red-Noise Null')
ax.axvspan(26.0, 28.5, color='gold', alpha=0.25, label='27-Day Solar Rotation Band')
ax.set_xlabel('Period (Days)')
ax.set_ylabel('Magnitude-Squared Coherence')
ax.set_title('Cross-Spectral Coherence: Oulu GCR Flux vs. WPWP Convective OLR')
ax.legend()
plt.tight_layout()
plt.savefig('../figures/fig03_red_noise_coherence.png', dpi=180)
plt.show()
```
